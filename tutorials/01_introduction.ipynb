# Introduction to vop_poc_nz

This tutorial demonstrates how to use the `vop_poc_nz` package to perform a Value of Perspective (VoP) analysis.

## 1. Setup
First, we import the necessary modules.

```python
import os
import yaml
import pandas as pd
import numpy as np
from vop_poc_nz.pipeline.analysis import run_analysis_pipeline
from vop_poc_nz.pipeline.reporting import run_reporting_pipeline
```

## 2. Load Configuration
We load the model parameters from a YAML file.

```python
config_path = "../src/parameters.yaml"
with open(config_path, "r") as f:
    config = yaml.safe_load(f)

# Let's look at the HPV Vaccination parameters
print(config["hpv_vaccination"]["states"])
```

## 3. Run Analysis
We run the analysis pipeline. This will execute the Markov models for both Health System and Societal perspectives, perform PSA, and calculate VoP metrics.

```python
output_dir = "tutorial_output"
results = run_analysis_pipeline(config, output_dir=output_dir)
```

## 4. Explore Results
The results dictionary contains nested results for each intervention.

```python
hpv_res = results["intervention_results"]["hpv_vaccination"]
print("Health System ICER:", hpv_res["health_system"]["icer"])
print("Societal ICER:", hpv_res["societal"]["icer"])
```

## 5. Generate Reports
Finally, we generate the figures and tables.

```python
run_reporting_pipeline(results, output_dir=output_dir)
print(f"Check {output_dir}/figures for the generated plots!")
```
